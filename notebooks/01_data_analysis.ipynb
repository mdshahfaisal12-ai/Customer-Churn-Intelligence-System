```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python
#Load the dataset by going up one directory first
data = pd.read_csv("../data/Telco-Customer-Churn.csv")

#Display the first 5 rows and data overview
print(f"Dataset Shape: {data.shape}\n")
data.head()
```

```python
#Check data types and missing values
print("--- Data Info ---")
data.info()

#View basic statistics for numerical columns
print("\n--- Summary Statistics ---")
data.describe()
```

```python
#Check the total count of Churn vs No Churn
print("--- Churn Counts ---")
print(data['Churn'].value_counts())

#Check the percentage of distribution
print("\n--- Churn Percentage ---")
print(data['Churn'].value_counts(normalize=True) * 100)
```

```python
#Set the visual style
sns.set(style="whitegrid")

#Create a figure canvas
plt.figure(figsize=(6, 4))

#Plot a bar chart showing churn count distributions
sns.countplot(x='Churn', data=data, hue='Churn', palette='Set2', legend=False)

#Add titles and labels to the chart
plt.title('Customer Churn Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Churn', fontsize=12)
plt.ylabel('Count', fontsize=12)

#Display the chart cleanly
plt.show()
```

```python
#Create a figure canvas
plt.figure(figsize=(10, 5))

#plot the distribution of Monthly Charge for Churn vs No Churn
sns.kdeplot(data=data, x='MonthlyCharges', hue='Churn', fill=True, common_norm=False, palette='Set2', alpha=0.5)

#Add clear descriptive labels
plt.title('Distribution of Monthly Charges by Churn Status', fontsize=14, fontweight='bold')
plt.xlabel('Monthly Charges ($)', fontsize=12)
plt.ylabel('Density', fontsize=12)

#Display the chart cleanly
plt.show()
```

```python
#Create a figure canvas
plt.figure(figsize=(10, 5))

#plot churn rate grouped by Contract type
sns.barplot(x='Contract', y='Churn', data=data, palette='Set2')

#Add titles and formatting
plt.title('Churn Rate by Contract Type', fontsize=14, fontweight='bold')
plt.xlabel('Contract Type', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.legend(title='Churned?')

#Display chart
plt.show()
```

```python
contract_churn = pd.crosstab(
    data['Contract'],
    data['Churn'],
    normalize='index'
) * 100

print(contract_churn.round(2))
```

```python
highest_churn_contract = contract_churn['Yes'].idxmax()
highest_churn_rate = contract_churn['Yes'].max()

print("Highest-risk contract:",
      highest_churn_contract, "with churn rate of", round(highest_churn_rate, 2), "%")
```
